```python
from agno.agent import Agent
from agno.models.groq import Groq # Import Groq instead of OpenAI
from agno.tools.yfinance import YFinanceTools


import os
from google.colab import userdata
from dotenv import load_dotenv

load_dotenv()
# Ensure your secret key in Colab is named exactly 'GROQ_API_KEY'
os.environ['GROQ_API_KEY'] = userdata.get('GROQ_API_KEY')

# Create the Finance Agent using Groq
finance_agent = Agent(
    name="Finance AI Agent",
    model=Groq(id="llama-3.3-70b-versatile"), # Use a fast Llama model via Groq
    tools=[YFinanceTools(enable_stock_price=True, enable_analyst_recommendations=True, enable_stock_fundamentals=True)],
    instructions=["Use tables to display the data."],
    markdown=True,
)

# Run the agent
finance_agent.print_response("Share the NVDA stock price and analyst recommendations", stream=True)
```
