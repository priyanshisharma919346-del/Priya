```python
from agno.agent import Agent
from agno.models.groq import Groq  # Fixed: Capital G

import os
from google.colab import userdata
from dotenv import load_dotenv

load_dotenv()
# Ensure your secret key in Colab is named exactly 'GROQ_API_KEY'
os.environ['GROQ_API_KEY'] = userdata.get('GROQ_API_KEY')


def build_agent():
    return Agent(
        # Qwen model ID yahan update karein
        model=Groq(id="llama-3.3-70b-versatile"),
        markdown=True,
        instructions=["You are a very helpful travel agent."],
        add_datetime_to_context=True
    )

# Execute
travel_agent = build_agent()
travel_agent.print_response("It is safe to travel UAE today ya other location like london")
```
